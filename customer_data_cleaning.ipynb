```python
import pandas as pd


path = "/Users/amalm/Downloads/dirty_customer_data_v2.csv"

df = pd.read_csv(path)
```

```python
print(df.shape)
print(df.columns.tolist())
```

```python
print(df.head(20))
#get an insight into how messy the data is and which columns need cleaning.
```

```python
#I can see country is written as DK and denmark so that needs to be corrected

print(df['Country'].unique())
```

```python
#there are 7 ways Denmark is written in the country column

df['Country'] = df['Country'].replace([ 'denmark', ' Denmark ', 'Danmark', 'DK', 'DANMARK', 'DENMARK' ], 'Denmark')
```

```python
print(df['Country'].unique())
```

```python
#Correct City

df['City'] = df['City'].str.capitalize()
```

```python
df['City'].unique()
```

```python
df['City'] = df['City'].replace(['openhagen', 'Copenhagen', 'copenhagen', 'københavn', 'københavn', ' copenhagen ', 'COPENHAGEN', ' Copenhagen ','København'], 'Copenhagen')
```

```python
df['City'] = df['City'].replace(['aarhus', 'århus', 'AARHUS', 'Århus'], 'Aarhus')
```

```python
df['City'] = df['City'].replace(['odense', ' odense ', ' Odense ','ODENSE'], 'Odense')
```

```python
df['City'] = df['City'].replace(['aalborg', 'ålborg', 'AALBORG', 'Ålborg'], 'Aalborg')
```

```python
df['City'].unique() 
```

```python
#Correct Signup_Date

df['Signup_Date'].dtype
```

```python
df['Signup_Date'] = pd.to_datetime(
    df['Signup_Date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
```

```python
print(df['Annual_Spend_DKK'].dtype)
```

```python
#remove DKK & , from Annual spend DKK
df['Annual_Spend_DKK'] = df['Annual_Spend_DKK'].str.replace('DKK', '')
df['Annual_Spend_DKK'] = df['Annual_Spend_DKK'].str.replace(',', '')
```

```python
df['Annual_Spend_DKK'] = pd.to_numeric(
   df['Annual_Spend_DKK'],
   errors= 'coerce' 
)
```

```python
print(df[df['Annual_Spend_DKK'] <0])
```

```python
#Find Annual_Spend_DKK
df.loc[df['Annual_Spend_DKK'] < 0, 'Annual_Spend_DKK'] = 0
```

```python
print("Missing dates:", df['Signup_Date'].isna().sum())
print("Missing spend:", df['Annual_Spend_DKK'].isna().sum())
```

```python
df['Age'].dtype
```

```python
df['Age'] = df['Age'].astype('Int64')
```

```python
#check age not > 100
print(df[df['Age'] > 100])
```

```python
#Find age < 0
print(df[df['Age'] < 0])
```

```python
# Convert age  < 0 to NaN
df.loc[df['Age'] < 0, 'Age'] = pd.NA
```

```python
# Convert age >100 to NaN
df.loc[df['Age'] >100, 'Age'] = pd.NA
```

```python
#Check to see if it has been changed
print(df[df['Age'] > 100])
print(df[df['Age'] < 0])
```

```python
df['Orders'].dtype
```

```python
#convert orders to integer to remove .0
df['Orders'] = df['Orders'].astype('Int64')
```

```python
#Find orders <0 
print(df[df['Orders'] <0])
```

```python
df.loc[df['Orders'] <0, 'Orders'] = pd.NA
```

```python
#Check to see orders < o have been changed
print(df[df['Orders'] <0])
```

```python
print("Min:", df['Orders'].min())
print("Max:", df['Orders'].max())
```

```python
#Email

df['Email'] = df['Email'].str.lower()

```

```python

# Show emails that do not contain @
df[df['Email'].notna() & ~df['Email'].str.contains('@')]
```

```python
# Emails without @ --> NaN
df.loc[
    df['Email'].notna() & ~df['Email'].str.contains('@'),
    'Email'
] = pd.NA
```

```python
#Check that there are no Email adresses left without @
df[df['Email'].notna() & ~df['Email'].str.contains('@')]
```

```python
#Find Emails withut UserID
df[df['Email'].notna() & ~df['Email'].str.contains('user')]
```

```python
# Emails without UserID --> NaN
df.loc[
    df['Email'].notna() & ~df['Email'].str.contains('user'),
    'Email'
] = pd.NA
```

```python
#check that there are no Email adresses left without userID 
df[df['Email'].notna() & ~df['Email'].str.contains('user')]
```
